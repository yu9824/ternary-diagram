# Scatter plot colored by values

```python
import matplotlib.pyplot as plt
import pandas as pd

from ternary_diagram import TernaryDiagram
```

```python
df = pd.read_csv("example_scatter.csv")
df.head()
```

```python
materials = df.columns[0:3]
materials
```

```python
fig, ax = plt.subplots(facecolor="w", layout="constrained")
td = TernaryDiagram(materials=materials, ax=ax)
td.scatter(df[materials], z=df["z"], z_min=0, z_max=1)
```

```python
td.fig.savefig("example_scatter.png", dpi=144)
```
