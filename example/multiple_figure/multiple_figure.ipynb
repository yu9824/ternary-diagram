# Multiple diagrams in one figure

Import modules

```python
import os

import pandas as pd
import matplotlib.pyplot as plt

from ternary_diagram import TernaryDiagram
```

Create the figure with `layout="constrained"`. Matplotlib's constrained layout keeps the diagrams, their labels and the colorbar from overlapping each other, so you don't need to adjust the layout afterwards or search for a large enough `figsize`. Pass each Axes to `TernaryDiagram` with `ax=`.

```python
fig, axes = plt.subplots(
    1, 2, dpi=72, facecolor="white", figsize=(10, 3.6), layout="constrained"
)
```

```python
fpath_example = ".."
fpath_contour = os.path.join(fpath_example, "contour", "example_contour.csv")
fpath_mono_scatter = os.path.join(
    fpath_example, "mono_scatter", "example_mono_scatter.csv"
)
```

```python
df_contour = pd.read_csv(fpath_contour)
df_mono_scatter = pd.read_csv(fpath_mono_scatter)
display(df_contour.head(), df_mono_scatter.head())
```

```python
materials = df_contour.columns[0:3]

td1 = TernaryDiagram(materials=materials, ax=axes[0])
td1.contour(vector=df_contour[materials], z=df_contour["z"])

fig
```

```python
materials = df_mono_scatter.columns[0:3]

td2 = TernaryDiagram(materials=materials, ax=axes[1])
td2.scatter(vector=df_mono_scatter[materials])

fig
```

```python
fig.savefig("multiple_figure.png", dpi=144)
```
