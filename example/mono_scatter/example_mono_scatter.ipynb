# Scatter plot and tie lines

```python
import matplotlib.pyplot as plt
import pandas as pd

from ternary_diagram import TernaryDiagram
```

```python
df = pd.read_csv("example_mono_scatter.csv")
df.head()
```

## Plot

```python
fig, ax = plt.subplots(facecolor="white", layout="constrained")
td = TernaryDiagram(df.columns, ax=ax)
td.scatter(df, color="blue", facecolor="None", lw=2)
```

## Tie line

```python
fig, ax = plt.subplots(facecolor="w", layout="constrained")
td = TernaryDiagram(df.columns, ax=ax)
td.scatter(df.iloc[:10], label="<10")
td.scatter(df.iloc[10:], label=">10")
td.plot([[1, 0, 1], [0, 2, 3]], label="tie line")
td.ax.legend()
```

```python
fig.savefig("example_mono_scatter.png", dpi=144)
```
